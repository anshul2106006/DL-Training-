```python
#Practical 2: Create  a Vocabulary

sentence = "I love deep learning"
tokens = sentence.split() # ['I','love','deep','learning']
# create a simple vocabulary
vocabulary ={
    word:index
    for index,word in enumerate(tokens,start=1)
}
#convert tokens into numerical IDs
token_ids =[
    vocabulary[word]
    for word in tokens
]
print("vocabulary:",vocabulary)
print("Token IDs",token_ids)
```

```python
# Practical 3: Create embeddings using TensorFlow

import tensorflow as tf
from tensorflow.keras import layers

# Four tokens in our small vocabulary
vocab_size = 5

# Every token will have an embedding of 3 numbers
embedding_dimension = 3

embedding_layer = layers.Embedding(
    input_dim=vocab_size,
    output_dim=embedding_dimension
)

# Token IDs for: I love deep learning
token_ids = tf.constant([[1, 2, 3, 4]])

#convert token IDs into embedding vectors
embeddings = embedding_layer(token_ids)

print("Token IDs shape:",token_ids.shape)
print("Embeddings shape:",embeddings.shape)
print(embeddings.numpy())
```

```python
#attention_practical

import tensorflow as tf

# Step 1: Define three query vectors
Q = tf.constant([
    [1.0, 0.0],
    [0.0, 1.0],
    [1.0, 1.0]
])

# Step 2: Define three key vectors
K = tf.constant([
    [1.0, 0.0],
    [0.0, 1.0],
    [1.0, 1.0]
])
#step 3: Define three value vectors
v=tf.constant([
    [10.0,0.0],
    [0.0,10.0],
    [5.0,5.0]
])
#step 4: Calculate query-key scores
scores = tf.matmul(Q,K,transpose_b=True)

#step 5: Scale the scores
key_dimension = tf.cast(
    tf.shape(K)[-1],
    tf.float32
)
scaled_scores = scores /tf.math.sqrt(key_dimension)

#step 6:Convert scores into attention weights
weights =tf.nn.softmax(scaled_scores,axis =-1)

#step 7: calculate the attention output
output = tf.matmul(weights,v)

print("Scores:\n",scores.numpy())
print("Scaled scores:\n",scaled_scores.numpy())
print("Attention weights:\n",weights.numpy())
print("Attention output:\n",output.numpy())
```
